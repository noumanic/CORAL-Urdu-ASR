```python
# Load Alif-1 model
from transformers import AutoTokenizer, AutoModelForCausalLM
import torch

model_id = "large-traversaal/Alif-1.0-8B-Instruct"
tokenizer = AutoTokenizer.from_pretrained(model_id)
model = AutoModelForCausalLM.from_pretrained(model_id, device_map="auto", torch_dtype=torch.float16)
```

```python
import ast
df_output = pd.read_csv('/kaggle/input/iteration-two-report/iteration_two_report_dataset.csv')

# Parse string representations back to lists of tuples
for m in ['whisper-large', 'whisper-medium', 'wav2vec2-urdu', 'whisper-small']:
    df_output[m] = df_output[m].apply(ast.literal_eval)

def get_alif_1_correction(row):
    # Gather all 4 model predictions with probabilities
    predictions = []
    for m in ['whisper-large', 'whisper-medium', 'wav2vec2-urdu', 'whisper-small']:
        if row[m]:
            words_probs = ', '.join([f"{w} ({p:.2f})" for w, p in row[m]])
            predictions.append(words_probs)
        else:
            predictions.append("(خالی)")
    
    print(f"predictions {predictions}")
    prompt = f"""یہ 4 ASR ماڈلز کی پیشن گوئیاں ہیں (لفظ اور اعتماد):
1. {predictions[0]}
2. {predictions[1]}
3. {predictions[2]}
4. {predictions[3]}

ایک صحیح جملہ اس طرح ہونا چاہیے: واضح، درست املا، مکمل الفاظ کے ساتھ۔
ان تمام پیشن گوئیوں کو مدنظر رکھتے ہوئے، صحیح ترین اردو جملہ لکھیں:"""
    
    inputs = tokenizer(prompt, return_tensors="pt").to(model.device)
    outputs = model.generate(
        **inputs, 
        max_new_tokens=100,
        temperature=0.5,
        do_sample=True,
        pad_token_id=tokenizer.eos_token_id
    )
    
    result = tokenizer.decode(outputs[0], skip_special_tokens=True)
    
    # Remove the prompt echo
    if "صحیح ترین اردو جملہ لکھیں:" in result:
        refined = result.split("صحیح ترین اردو جملہ لکھیں:")[-1].strip()
    else:
        refined = result[len(prompt):].strip()
    
    # If still empty, use best prediction (whisper-large)
    if not refined:
        words = [w for w, p in row['whisper-large']] if row['whisper-large'] else []
        refined = ' '.join(words)
    
    return refined

# WER calculation
def wer(ref, hyp):
    
    ref_w = ref.split()
    hyp_w = hyp.split() if hyp else []
    
    # Handle empty cases
    if len(ref_w) == 0:
        return 0.0 if len(hyp_w) == 0 else 1.0
    
    # Levenshtein distance
    d = [[0]*(len(hyp_w)+1) for _ in range(len(ref_w)+1)]
    
    for i in range(len(ref_w)+1):
        d[i][0] = i
    for j in range(len(hyp_w)+1):
        d[0][j] = j
    
    for i in range(1, len(ref_w)+1):
        for j in range(1, len(hyp_w)+1):
            if ref_w[i-1] == hyp_w[j-1]:
                d[i][j] = d[i-1][j-1]
            else:
                substitution = d[i-1][j-1] + 1
                insertion = d[i][j-1] + 1
                deletion = d[i-1][j] + 1
                d[i][j] = min(substitution, insertion, deletion)
    
    # WER = edit_distance / reference_length
    wer_score = d[len(ref_w)][len(hyp_w)] / len(ref_w)
    
    # Cap at 1.0 (shouldn't exceed 100% but can in some implementations)
    return min(wer_score, 1.0)

alif_results = []
wer_scores = []

for idx, row in df_output.iterrows():
    # Get Alif-1 prediction
    alif_pred = get_alif_1_correction(row)
    alif_results.append(alif_pred)
    
    # Calculate WER
    print(f"ground truth: {row['sentence']}")
    print(f"alif_1: {alif_pred}")

    wer_score = wer(row['sentence'], alif_pred)
    wer_scores.append(wer_score)
    
    print(f"WER: {wer_score:.4f}")
    print("-" * 80 + "\n")

# Add to dataframe
df_output['alif_1'] = alif_results
df_output['WER'] = wer_scores

print(f"Average WER: {df_output['WER'].mean():.4f}")
df_output.to_csv('results_with_alif.csv', index=False)
print("\nSaved to results_with_alif.csv")
```
